```python
# process 3d 1mm data to 1d 2mm data
import os
import pandas as pd
import nibabel as nib
import numpy as np
import nilearn.plotting

save_root = "./visualize"

mask_path = './labels/MNI152_T1_2mm_brain.nii.gz'
mask_img = nib.load(mask_path)
mask_data = mask_img.get_fdata()
mask_nonzero = (mask_data != 0)
```

```python
data_path = "./demo_dataset/sub_1/T1_brain_2mm_stdspace.npy"
image = np.load(data_path).reshape(-1)
image_3d = mask_data.copy()
lower, upper = np.percentile(image, 0.05), np.percentile(image, 99.95)
image = np.clip(image,lower,upper)
image_3d[mask_nonzero] = (image -image.min())/(image.max() -image.min())
gen_img = nib.Nifti1Image(image_3d, affine=mask_img.affine, header=mask_img.header)
output_path = os.path.join(save_root,'corrupted_images.nii.gz')
nib.save(gen_img, output_path)
nilearn.plotting.plot_anat(output_path,display_mode='z',cut_coords=range(-40,41,10), black_bg=True)
```
